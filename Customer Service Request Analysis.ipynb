# Step 2: Exploratory Data Analysis

```python
# Frequency Plot of null values
df.isnull().sum().plot(kind = 'bar', figsize = (12,6))
plt.title('Null values count')
plt.show()
```

```python
# Missing Value Treatments
# Removes rows where closed date is null
```

```python
df = df[df['Closed Date'].notnull()]
```

```python
df['Closed Date'].isnull().sum()
```

```python
df.isnull().sum()
```

```python
# Analyze the date column
df = df.copy()
df['Created Date']=pd.to_datetime(df['Created Date'], errors = 'coerce')
df['Closed Date']=pd.to_datetime(df['Closed Date'], errors = 'coerce')
```

```python
df[['Created Date', 'Closed Date']].dtypes
```

```python
df[['Created Date', 'Closed Date']].head()
```

```python
# Remove Incorrect Timeline
df = df[df['Closed Date'] >= df['Created Date']]
```

```python
df[['Created Date', 'Closed Date']].head()
```

```python
# Calculate Time Elapsed in Closed & Creation Date
df['Request Closing Time'] = df['Closed Date'] - df['Created Date']
```

```python
df.shape
```

```python
df.columns
```

```python
# Convert Calculated Date to Seconds
df['Request Closing Time'] = df['Request Closing Time'].dt.total_seconds()
```

# Step 6: Statistical Analysis using P-Values

```python
from scipy.stats import f_oneway
groups = [group['Request Closing Time'].values
          for name, group in df.groupby('Complaint Type')]
f_stat, p_value = f_oneway(*groups)
print(p_value)
```

# Step 7 : Kruskal-Wallis Test

```python
from scipy.stats import kruskal

groups = [group['Request Closing Time'].values 
          for name, group in df.groupby('Complaint Type')]

stat, p = kruskal(*groups)

print(p)
```
